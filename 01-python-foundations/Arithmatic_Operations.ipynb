```python
print("# Python Arithmetic Operators Tutorial for Beginners")
print("\nThis tutorial will guide you through the basic arithmetic operators in Python.")

print("\n---")
print("# 1. Addition (+)")
print("The addition operator adds two operands.")
# Example for Addition
num1 = 10
num2 = 5
result_add = num1 + num2
print(f"Example: {num1} + {num2} = {result_add}") # Output: 15

print("\n---")
print("# 2. Subtraction (-)")
print("The subtraction operator subtracts the second operand from the first.")
# Example for Subtraction
result_sub = num1 - num2
print(f"Example: {num1} - {num2} = {result_sub}") # Output: 5

print("\n---")
print("# 3. Multiplication (*)")
print("The multiplication operator multiplies two operands.")
# Example for Multiplication
result_mul = num1 * num2
print(f"Example: {num1} * {num2} = {result_mul}") # Output: 50

print("\n---")
print("# 4. Division (/)")
print("The division operator divides the first operand by the second. It always returns a float.")
# Example for Division
result_div = num1 / num2
print(f"Example: {num1} / {num2} = {result_div}") # Output: 2.0

print("\n---")
print("# 5. Modulus (%)")
print("The modulus operator returns the remainder when the first operand is divided by the second.")
# Example for Modulus
num3 = 10
num4 = 3
result_mod = num3 % num4
print(f"Example: {num3} % {num4} = {result_mod}") # Output: 1 (10 divided by 3 is 3 with a remainder of 1)

print("\n---")
print("# 6. Exponentiation (**)")
print("The exponentiation operator raises the first operand to the power of the second operand.")
# Example for Exponentiation
base = 2
power = 3
result_exp = base ** power
print(f"Example: {base} ** {power} = {result_exp}") # Output: 8 (2*2*2)

print("\n---")
print("# 7. Floor Division (//)")
print("The floor division operator divides and rounds down to the nearest integer. The result is an integer.")
# Example for Floor Division
num5 = 10
num6 = 3
result_floor_div = num5 // num6
print(f"Example: {num5} // {num6} = {result_floor_div}") # Output: 3 (10 divided by 3 is 3.33, rounded down to 3)

print("\nFor negative numbers, floor division rounds down to the more negative integer.")
negative_num = -10
result_floor_neg = negative_num // num6
print(f"Example: {negative_num} // {num6} = {result_floor_neg}") # Output: -4 ( -3.33 rounded down to -4)

print("\n---")
print("\nPractice these operators to get a good grasp of basic arithmetic in Python!")
```
