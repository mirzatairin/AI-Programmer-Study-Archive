# Python Operators

Common arithmetic, assignment, comparison, and logical operators.

## 1. Arithmetic Operators

Python supports addition, subtraction, multiplication, division, modulus, exponentiation, and floor division.

```python
a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a % b)   # Remainder
print(a ** b)  # Exponentiation
print(a // b)  # Floor division
```

## 2. Assignment Operators

Assignment operators update variables.

```python
score = 10
score += 5
print(score)

score -= 2
print(score)

score *= 2
print(score)
```

## 3. Comparison Operators

Comparison expressions produce `True` or `False`.

```python
x = 10
y = 20

print(x < y)
print(x > y)
print(x <= y)
print(x >= y)
print(x == y)
print(x != y)
```

## 4. Logical Operators

`and`, `or`, and `not` combine or invert Boolean expressions.

```python
age = 20
has_id = True

print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

## 5. Useful Built-in Functions

A few common functions are `abs()`, `round()`, `min()`, `max()`, `sum()`, and `len()`.

```python
numbers = [4, 8, 2, 10]

print(abs(-7))
print(round(3.14159, 2))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(len(numbers))
```
