```python
##Practice Problem: Using list indexing to find the price of a product
#Product data in the format (product_name, price)
products_data = [("Laptop", 1200), ("Mouse", 25), ("Keyboard", 75), ("Monitor", 300), ("Webcam", 50)]

# The index of the product for which we want to find the price
target_product_index = 1

# TODO: replace None with appropriate code
# Use list indexing to determine the price for `target_product_index`
product_price = products_data[target_product_index][1]

### Notebook grading
correct_answer = 25
if product_price != correct_answer:
    print("Not quite! Did you correctly access the price element of the product at the given index?")
else:
    print(f"Nice work! I found the price like this: `products_data[target_product_index][1]`.")
```

```python
## Quiz: Palindrome Check on Image Classifications

#in this exercise, you'll use list slicing to determine if a sequence of image classifications forms a palindrome. A palindrome is a sequence that reads the same backward as forward.

#The list image_classifications contains the predicted classes of a series of images processed by an image classification model.

#Hint: You need to use list slicing to reverse the list. The slicing syntax `[::-1]` reverses the list.



# Predicted classes of a series of images
dessert_classifications = ['icecream', 'smoothie', 'chocolate', 'smoothie', 'icecream']

# TODO: replace None with appropriate code
# Use list slicing to reverse the list and check if it forms a palindrome
if (dessert_classifications == dessert_classifications[::-1]):
    is_palindrome = True
else:
    is_palindrome = False


### Notebook grading
correct_answer = True
if is_palindrome != correct_answer:
    print("Not quite! Remember to use list slicing to reverse the list and check if it matches the original list.")
else:
    print("Nice work! I found is_palindrome like this: `dessert_classifications == dessert_classifications[::-1]`.")
```

```python
#Tuple packing and unpacking example

dimensions = 52, 40, 100 #This line creates a tuple named dimensions containing three integer values: 52, 40, and 100. In Python, you don't always need parentheses to create a tuple; simply separating values with commas works.
length, width, height = dimensions #This is an example of tuple unpacking. It takes the elements from the dimensions tuple and assigns them to corresponding variables on the left side of the assignment operator. So, length becomes 52, width becomes 40, and height becomes 100.
print("The dimensions are {} x {} x {}".format(length, width, height)) #This line prints a formatted string to the console. The {} curly braces are placeholders that will be replaced by the values provided to the .format() method, in the order they are given. Therefore, it will output: "The dimensions are 52 x 40 x 100".
```

```python
import time

# Sample data
set1 = set(range(1000))
set2 = set(range(500, 1500))
list1 = list(set1)
list2 = list(set2)

# Union
start_time = time.time()
union_set = set1.union(set2)
print("Set Union Time:", time.time() - start_time)

start_time = time.time()
union_list = list(set(list1 + list2))
print("List Union Time:", time.time() - start_time)

# Intersection
start_time = time.time()
intersection_set = set1.intersection(set2)
print("Set Intersection Time:", time.time() - start_time)

start_time = time.time()
intersection_list = [x for x in list1 if x in set2]
print("List Intersection Time:", time.time() - start_time)

# Difference
start_time = time.time()
difference_set = set1.difference(set2)
print("Set Difference Time:", time.time() - start_time)

start_time = time.time()
difference_list = [x for x in list1 if x not in set2]
print("List Difference Time:", time.time() - start_time)
```
