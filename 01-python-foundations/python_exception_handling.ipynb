# Python Exceptions

Common beginner-level exceptions and basic exception handling.

## 1. Common Python Errors

Errors are normal while programming. Learning to identify them is an important debugging skill.

### ZeroDivisionError

Occurs when a number is divided by zero.

```python
# Uncomment to see the error:
# result = 10 / 0
```

### ValueError

Occurs when a value has the right general type but cannot be converted or interpreted as requested.

```python
# Uncomment to see the error:
# number = int("hello")
```

### TypeError

Occurs when an operation is used with an inappropriate type.

```python
# Uncomment to see the error:
# result = "10" + 5
```

## 2. Handling Exceptions with `try` and `except`

Use `try` for code that might fail and `except` to handle a known error.

```python
user_input = "hello"

try:
    number = int(user_input)
    print(number)
except ValueError:
    print("Please provide a valid integer.")
```

## 3. Handling Different Exceptions

You can use more than one `except` block when different errors need different handling.

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ValueError:
    print("Please enter a valid integer.")
except ZeroDivisionError:
    print("The number cannot be zero.")
```
