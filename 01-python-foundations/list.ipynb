```python
# --- Python Lists: Rin's Tutorial ---

# Introduction to Lists
# Lists are ordered collections of items. They are one of the most versatile data types in Python.
# Think of a list as a container that can hold various types of data, such as numbers, strings, or even other lists.

# Key characteristics of Python lists:
# 1. Ordered: The order of items in a list is maintained.
# 2. Mutable: You can change, add, or remove items after the list is created.
# 3. Allows duplicates: A list can contain the same item multiple times.
# 4. Heterogeneous: A list can contain items of different data types (though often, lists contain items of the same type).

# 1. Creating a List
# Lists are created using square brackets `[]` with items separated by commas.
desserts = ["cookies", "strawberryshortcake", "muffins"]
print("Original list of desserts:", desserts)
print("-" * 30)

# 2. Accessing Elements (Indexing)
# List items are accessed by their index. Python uses 0-based indexing,
# meaning the first item is at index 0, the second at index 1, and so on.
print("First dessert (index 0):", desserts[0])
print("Third dessert (index 2):", desserts[2])

# Negative indexing allows access from the end of the list.
# -1 refers to the last item, -2 to the second to last, etc.
print("Last dessert (index -1):", desserts[-1])
print("-" * 30)

# 3. Getting the Length of a List
# The `len()` function returns the number of items in a list.
print("Number of desserts in the list:", len(desserts))
print("-" * 30)

# 4. Modifying Elements
# Since lists are mutable, you can change the value of an item at a specific index.
print("List before modification:", desserts)
desserts[2] = 'cheesecake' # Changing 'muffins' to 'cheesecake'
print("List after modifying item at index 2:", desserts)
print("-" * 30)

# 5. Adding Elements to a List
# a. `append()`: Adds an item to the end of the list.
desserts.append('chocolate cheesecake')
print("List after appending 'chocolate cheesecake':", desserts)

# b. `insert(index, item)`: Inserts an item at a specified index.
desserts.insert(1, 'ice cream') # Insert 'ice cream' at index 1
print("List after inserting 'ice cream' at index 1:", desserts)
print("-" * 30)

# 6. Removing Elements from a List
# a. `remove(item)`: Removes the first occurrence of a specified item.
#    Raises a ValueError if the item is not found.
print("List before removing 'cookies':", desserts)
desserts.remove('cookies')
print("List after removing 'cookies':", desserts)

# b. `pop(index)`: Removes and returns the item at a given index.
#    If no index is specified, `pop()` removes and returns the last item.
popped_dessert = desserts.pop() # Removes the last item
print("List after popping the last item:", desserts)
print("Popped item:", popped_dessert)

popped_dessert_at_index_0 = desserts.pop(0) # Removes item at index 0
print("List after popping item at index 0:", desserts)
print("Popped item at index 0:", popped_dessert_at_index_0)
print("-" * 30)

# 7. Common Pitfall: Index Errors
# Attempting to access an index that does not exist in the list will raise an `IndexError`.
# This is why understanding `len()` is important!
print("Current desserts list:", desserts)
print("Length of list:", len(desserts))

# The list currently has items at indices 0, 1.
# Attempting to access an index like 25 would cause an error.
try:
    print("Attempting to access index 25:", desserts[25])
except IndexError as e:
    print(f"Caught an error: {e}. This means the index 25 is out of bounds for the current list.")

# Always check if an index is valid before accessing:
if len(desserts) > 1: # Check if there's at least an item at index 1
    print("Item at index 1 (safe access):", desserts[1])
else:
    print("List is too short to access index 1 safely.")
print("-" * 30)

# 8. List Slicing (Optional but useful for beginners)
# You can extract a portion of a list using slicing: `list[start:end:step]`
# `start` is inclusive, `end` is exclusive.
my_list = [10, 20, 30, 40, 50]
print("Original list for slicing:", my_list)
print("Items from index 1 to 3 (exclusive):", my_list[1:4]) # Gets 20, 30, 40
print("Items from beginning to index 2 (exclusive):", my_list[:3]) # Gets 10, 20, 30
print("Items from index 2 to end:", my_list[2:]) # Gets 30, 40, 50
print("A copy of the list:", my_list[:])
print("-" * 30)

# Summary of Important Lessons:
# - Lists are ordered and mutable collections.
# - Use square brackets `[]` to define a list.
# - Access elements using 0-based indexing `list[index]`.
# - Negative indexing `list[-1]` accesses from the end.
# - Use `len(list)` to get the number of elements.
# - Modify elements by assigning `list[index] = new_value`.
# - Add elements with `append()` (end) or `insert(index, item)` (specific position).
# - Remove elements with `remove(item)` (by value) or `pop(index)` (by index).
# - Be careful of `IndexError` when accessing elements; ensure the index is valid.
```

```python
#we can also nest a list
Cats = ['meows', ['very cute', 'may bite'], 'soft', 'small']
print(Cats)
print(Cats[1])
print(Cats[1][0])

#lists can also have different types of variables stored in them
things = ["apple", 3.14, True]
print(things)

coordinates = (23.8, 90.4)
print(coordinates)
```
