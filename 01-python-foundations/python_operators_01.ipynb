```python

### Python Operators
# Common arithmetic, assignment, comparison, and logical operators.

a = 10
b = 3

print(a + b)   # Addition, prints 13
print(a - b)   # Subtraction, prints 7
print(a * b)   # Multiplication, prints 30
print(a / b)   # Division, prints 3.3333
print(a % b)   # Remainder, prints 1
print(a ** b)  # Exponentiation, prints 1000
print(a // b)  # Floor division, prints 3

## 2. Assignment Operators
# Assignment operators update variables.


score = 10
score += 5
print(score) # prints 15

score -= 2
print(score) # prints 13

score *= 2
print(score) #prints 26

#3. Comparison Operators
#Comparison expressions produce True or False.


x = 10
y = 20

print(x < y) #True
print(x > y) #False
print(x <= y) #True
print(x >= y) #False
print(x == y) #False
print(x != y) #True

# 4. Logical Operators
#and, or, and not combine or invert Boolean expressions.


age = 20
has_id = True

print(age >= 18 and has_id) #True
print(age < 18 or has_id) #True
print(not has_id) #False

#5. Useful Built-in Functions
#A few common functions are abs(), round(), min(), max(), sum(), and len().


numbers = [4, 8, 2, 10]

print(abs(-7)) #prints 7
print(round(3.14159, 2)) #prints 3.14
print(min(numbers)) #prints 2
print(max(numbers)) #prints 10
print(sum(numbers)) #prints 24
print(len(numbers)) #prints 4
```
