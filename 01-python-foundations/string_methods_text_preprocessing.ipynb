## Combining Text Preprocessing and Result Formatting

This section demonstrates how to first clean and prepare text data using string methods, and then how to present insights or results from that data using string formatting. These are fundamental steps in many Natural Language Processing (NLP) tasks.

```python
# Original raw text data for an NLP task
raw_review = "    This product is AMAZING! I love it. Highly recommended to everyone. "

print(f"Original text: '{raw_review}'")
```

### Step 1: Text Preprocessing

We'll clean the `raw_review` by:
1.  Removing leading/trailing whitespace using `.strip()`.
2.  Converting to lowercase using `.lower()` to ensure consistency.
3.  Replacing common contractions or symbols (if any) - for this example, we'll replace an exclamation mark.
4.  Splitting the processed text into individual words.

```python
# 1. Remove leading/trailing whitespace
cleaned_text = raw_review.strip()
print(f"After strip(): '{cleaned_text}'")

# 2. Convert to lowercase
cleaned_text = cleaned_text.lower()
print(f"After lower(): '{cleaned_text}'")

# 3. Replace specific punctuation (simple example)
cleaned_text = cleaned_text.replace('!', '')
print(f"After replace('!',''): '{cleaned_text}'")

# 4. Split the text into words
words_list = cleaned_text.split()
print(f"Words list: {words_list}")
```

### Step 2: Extracting Information and Formatting Results

After preprocessing, we can extract some simple information, like the number of words, and then present it in a human-readable format using `.format()` or f-strings.

```python
# Calculate a simple metric: the number of words
word_count = len(words_list)

# Format a result sentence using the extracted information
# We'll use an f-string here for modern Python formatting, which is similar to .format()
result_message = f"The analyzed review had {word_count} words after preprocessing. The first few words are: {words_list[:3]}."

print(result_message)
```
