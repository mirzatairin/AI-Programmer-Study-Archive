# Python Strings

Common string operations and a small introduction to text analysis.

## 1. Creating Strings

Strings can use single or double quotes.

```python
first_name = "Rin"
message = 'Hello, Python!'

print(first_name)
print(message)
```

## 2. Length and Indexing

Use `len()` to find the number of characters. Python uses zero-based indexing.

```python
text = "Python"

print(len(text))
print(text[0])
print(text[-1])
```

## 3. Common String Methods

Useful fundamentals include `lower()`, `upper()`, `capitalize()`, `replace()`, `split()`, `find()`, `count()`, `startswith()`, and `endswith()`.

```python
text = "Python is Fun"

print(text.lower())
print(text.upper())
print(text.capitalize())
print(text.replace("Fun", "powerful"))
print(text.split())
print(text.find("Python"))
print(text.count("o"))
print(text.startswith("Python"))
print(text.endswith("Fun"))
```

## 4. Joining Strings

`join()` combines multiple strings using a separator.

```python
words = ["Python", "for", "AI"]

sentence = " ".join(words)
print(sentence)
```

## 5. Formatting Strings

F-strings are a clean way to insert variables into text.

```python
name = "Rin"
score = 95

message = f"{name} scored {score}."
print(message)
```

## 6. Simple Text Analysis

Basic string operations can help with introductory NLP tasks such as counting words or preparing text for tokenization.

```python
text = "I love learning Python and I love building AI projects."

words = text.lower().split()
print(words)
print("Number of tokens:", len(words))
print("Occurrences of love:", words.count("love"))
```
