# Python Data Types

Common Python data types and basic type conversion.

## 1. Integers and Floats

Integers (`int`) are whole numbers, whether positive or negative. Floats (`float`) represent numbers with decimal values.
https://www.youtube.com/watch?v=ppsCxnNm-JI&t=2s

```python
# Integer
age = 28
print(age)
print(type(age))
```

```python
# Float
height = 1.65
print(height)
print(type(height))
```

## 2. Booleans

A Boolean (`bool`) can have one of two values: `True` or `False`.

```python
is_learning = True
is_finished = False

print(is_learning)
print(type(is_learning))
```

## 3. Strings

A string (`str`) is a sequence of characters enclosed in quotes.

```python
name = "Rin"
message = "Learning Python for AI"

print(name)
print(message)
print(type(message))

print(name*3)
print(message[:])
print(message[:5])
print(message[2:])
print(message[0:5])
print(message[0])
```

```python
# How to reverse a string

my_string = "Hello Python"
print(f"Original string: {my_string}")

# Method 1: Using slicing (most common and Pythonic way)
reversed_string_slice = my_string[::-1]
print(f"Reversed using slicing: {reversed_string_slice}")

# Method 2: Using the reversed() function and join()
# reversed() returns an iterator, so we need to join its elements
reversed_string_join = "".join(reversed(my_string))
print(f"Reversed using reversed() and join(): {reversed_string_join}")

# Method 3: Using a loop (more verbose, but good for understanding)
reversed_string_loop = ""
for char in my_string:
    reversed_string_loop = char + reversed_string_loop
print(f"Reversed using a loop: {reversed_string_loop}")
```

## 4. Checking Types

Use `type()` to inspect the data type of a value.

```python
x = 42
y = 3.14
z = True
text = "Python"

print(type(x))
print(type(y))
print(type(z))
print(type(text))
```

## 5. Type Conversion

Common conversions include `int()`, `float()`, `str()`, and `bool()`.

```python
number_text = "25"
number = int(number_text)
print(number, type(number))

price = 19
price_as_float = float(price)
print(price_as_float, type(price_as_float))

score = 95
score_text = str(score)
print(score_text, type(score_text))
```

## 6. Lists, Tuples, and Dictionaries

These are common Python collection types. The biggest differnce between a list and a tuple is that a tuple is called something as "immutable". it means it can not modified or changed after it has been created. A list is ordered and mutable, a tuple is ordered and immutable, and a dictionary stores key-value pairs.

```python
languages = ["Python", "Java", "C++"]
print(languages)

student = {"name": "Rin", "major": "CSE"}
print(student)
print(student["name"])
```

```python
desserts = ["cookies", "strawberryshortcake", "muffins"]
print(desserts)

#we can add an item to lists
desserts.append('chocolate cheesecake')
print(desserts)

#we can print the items separately
print(desserts[0])
print(desserts[2])

#we can remove an item
desserts.remove('cookies')
print(desserts)

#we can also change the value of an item in a list
desserts[2] = 'cheesecake'
print(desserts)
```

```python
#we can also nest a list
Cats = ['meows', ['very cute', 'may bite'], 'soft', 'small']
print(Cats)
print(Cats[1])
print(Cats[1][0])

#lists can also have different types of variables stored in them
things = ["apple", 3.14, True]
print(things)

coordinates = (23.8, 90.4)
print(coordinates)
```

```python
#A tuple example
tuple_example = ("apple", "banana", "cherry")
print(tuple_example)

#we can not add or remove items in tuple
```

```python
print("\n--- Python Sets Example ---")

# Creating a set, for a set we use these curly brackets
my_set = {"apple", "banana", "cherry"}
print(f"Original set: {my_set}")
print(f"Type of my_set: {type(my_set)}")

# Sets do not allow duplicate values
duplicate_set = {"apple", "banana", "cherry", "apple"}
print(f"Set with duplicates (duplicates are automatically removed): {duplicate_set}")

# Adding an element to a set
my_set.add("orange")
print(f"Set after adding 'orange': {my_set}")

# Removing an element from a set
my_set.remove("banana")
print(f"Set after removing 'banana': {my_set}")

# Another way to remove, but raises an error if item is not present
# my_set.remove("grape") # This would cause an error

# Discarding an element (does not raise an error if item is not present)
my_set.discard("grape")
print(f"Set after discarding 'grape' (no change as 'grape' was not present): {my_set}")
my_set.discard("apple")
print(f"Set after discarding 'apple': {my_set}")

# Set operations
set_a = {"apple", "banana", "cherry", "date"}
set_b = {"cherry", "date", "elderberry", "fig"}

# Union (all unique elements from both sets)
union_set = set_a.union(set_b)
print(f"Union of set_a and set_b: {union_set}")

# Intersection (common elements in both sets)
intersection_set = set_a.intersection(set_b)
print(f"Intersection of set_a and set_b: {intersection_set}")

# Difference (elements in set_a but not in set_b)
difference_set = set_a.difference(set_b)
print(f"Difference (set_a - set_b): {difference_set}")

# Symmetric Difference (elements in either set, but not in both)
symmetric_difference_set = set_a.symmetric_difference(set_b)
print(f"Symmetric Difference: {symmetric_difference_set}")
```

```python
print("\n--- Python Dictionaries Tutorial ---")

# What is a Dictionary?
# A dictionary is a collection that stores data in 'key: value' pairs.
# Think of it like a real-world dictionary where each word (key) has a definition (value).
# Dictionaries are unordered, changeable, and do not allow duplicate keys.

# 1. Creating a Dictionary

# An empty dictionary
empty_dict = {}
print(f"Empty dictionary: {empty_dict}")
print(f"Type of empty_dict: {type(empty_dict)}")

# A dictionary with some initial data
student_profile = {
    "name": "Rin",
    "age": 28,
    "major": "Computer Science",
    "is_enrolled": True
}
print(f"\nStudent profile: {student_profile}")

# Keys must be unique! If you add a key that already exists, it will update the value.
duplicate_key_dict = {"color": "red", "fruit": "apple", "color": "blue"}
print(f"Dictionary with duplicate key (last value wins): {duplicate_key_dict}")

# 2. Accessing Values
# We can access values using their keys inside square brackets.

print(f"\nStudent's name: {student_profile["name"]}")
print(f"Student's age: {student_profile["age"]}")

# We can also use the .get() method. It's safer because it won't give an error
# if the key doesn't exist, returning None instead (or a default value you specify).
print(f"Student's major (using .get()): {student_profile.get("major")}")
print(f"Student's GPA (key not found, returns None): {student_profile.get("GPA")}")
print(f"Student's GPA (key not found, returns 'N/A'): {student_profile.get("GPA", "N/A")}")

# 3. Adding and Modifying Elements
# To add a new key-value pair, just assign a value to a new key.
# To change an existing value, assign a new value to an existing key.

student_profile["year"] = "Senior" # Add a new key-value pair
print(f"\nProfile after adding 'year': {student_profile}")

student_profile["age"] = 29 # Modify an existing value
print(f"Profile after changing 'age': {student_profile}")

# 4. Removing Elements

# Using del keyword to remove an item by its key
del student_profile["is_enrolled"]
print(f"\nProfile after deleting 'is_enrolled': {student_profile}")

# Using .pop() method to remove an item by its key and return its value
removed_major = student_profile.pop("major")
print(f"Profile after popping 'major': {student_profile}")
print(f"Removed major was: {removed_major}")

# Using .clear() to remove all items from the dictionary
student_profile.clear()
print(f"Profile after clearing all items: {student_profile}")

# Let's re-create a dictionary for more examples
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(f"\nNew car dictionary: {car}")

# 5. Looping Through a Dictionary

# Loop through keys (this is the default behavior)
print("\nLooping through keys:")
for key in car:
    print(key)

# Loop through keys explicitly
print("\nLooping through keys explicitly:")
for key in car.keys():
    print(key)

# Loop through values
print("\nLooping through values:")
for value in car.values():
    print(value)

# Loop through both keys and values (items)
print("\nLooping through items (key: value):")
for key, value in car.items():
    print(f"{key}: {value}")

# 6. Checking if a Key Exists
# Use the 'in' keyword to check if a key is present in the dictionary.

print(f"\nIs 'model' in car dictionary? {'model' in car}")
print(f"Is 'color' in car dictionary? {'color' in car}")
print(f"Is 'engine' not in car dictionary? {'engine' not in car}")

# 7. Dictionary Length
# Use len() to find the number of key-value pairs.
print(f"\nNumber of items in car dictionary: {len(car)}")
```
