# Python Data Types

Common Python data types and basic type conversion.

## 1. Integers and Floats

Integers (`int`) are whole numbers. Floats (`float`) represent numbers with decimal values.

```python
# Integer
age = 28
print(age)
print(type(age))

# Float
height = 1.65
print(height)
print(type(height))
```

## 2. Booleans

A Boolean (`bool`) can have one of two values: `True` or `False`.

```python
is_learning = True
is_finished = False

print(is_learning)
print(type(is_learning))
```

## 3. Strings

A string (`str`) is a sequence of characters enclosed in quotes.

```python
name = "Rin"
message = "Learning Python for AI"

print(name)
print(message)
print(type(message))
```

## 4. Checking Types

Use `type()` to inspect the data type of a value.

```python
x = 42
y = 3.14
z = True
text = "Python"

print(type(x))
print(type(y))
print(type(z))
print(type(text))
```

## 5. Type Conversion

Common conversions include `int()`, `float()`, `str()`, and `bool()`.

```python
number_text = "25"
number = int(number_text)
print(number, type(number))

price = 19
price_as_float = float(price)
print(price_as_float, type(price_as_float))

score = 95
score_text = str(score)
print(score_text, type(score_text))
```

## 6. Lists, Tuples, and Dictionaries

These are common Python collection types. A list is ordered and mutable, a tuple is ordered and immutable, and a dictionary stores key-value pairs.

```python
languages = ["Python", "Java", "C++"]
print(languages)

coordinates = (23.8, 90.4)
print(coordinates)

student = {"name": "Rin", "major": "CSE"}
print(student)
print(student["name"])
```
