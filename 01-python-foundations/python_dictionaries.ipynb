```python
print("\n--- Python Dictionaries Tutorial ---")

# What is a Dictionary in Python?
# A dictionary is a collection that stores data in 'key: value' pairs.
# Dictionaries are unordered, mutable, and do not allow duplicate keys.

# 1. Creating a Dictionary

# An empty dictionary
empty_dict = {}
print(f"Empty dictionary: {empty_dict}") # Empty dictionary: {}
print(f"Type of empty_dict: {type(empty_dict)}") # Type of empty_dict: <class 'dict'>

# A dictionary with some initial data
student_profile = {
    "name": "Rin",
    "age": 28,
    "major": "Computer Science",
    "is_enrolled": True
}
print(f"\nStudent profile: {student_profile}")
#Student profile: {'name': 'Rin', 'age': 28, 'major': 'Computer Science', 'is_enrolled': True}

# Keys must be unique!!! If you add a key that already exists, it will simply update the value.
duplicate_key_dict = {"color": "red", "fruit": "apple", "color": "blue"}
print(f"Dictionary with duplicate key (last value wins): {duplicate_key_dict}") # Dictionary with duplicate key (last value wins): {'color': 'blue', 'fruit': 'apple'}


# 2. Accessing Values
# We can access values using their keys inside square brackets.

print(f"\nStudent's name: {student_profile["name"]}") #Student's name: Rin
print(f"Student's age: {student_profile["age"]}")#Student's age: 28

# We can also use the .get() method. But why use the .get() method??
# because It's safer, it won't give an error if the key doesn't exist, returning None instead (or a default value you specify).
print(f"Student's major (using .get()): {student_profile.get("major")}")#Student's major (using .get()): Computer Science
print(f"Student's GPA (key not found, returns None): {student_profile.get("GPA")}")
#Notice how "GPA" was never an existing key in our original dictionary? which is why,
#the output will be: Student's GPA (key not found, returns None): None
print(f"Student's GPA (key not found, returns 'N/A'): {student_profile.get("GPA", "N/A")}")#Student's GPA (key not found, returns 'N/A'): N/A
#In this case, we get N/A in stead of None cos we speicified that!


# 3. Adding and Modifying Elements
student_profile["year"] = "Senior" # We add a new key-value pair
print(f"\nProfile after adding 'year': {student_profile}")
#Profile after adding 'year': {'name': 'Rin', 'age': 28, 'major': 'Computer Science', 'is_enrolled': True, 'year': 'Senior'}

student_profile["age"] = 29 # Modify an existing value
print(f"Profile after changing 'age': {student_profile}")
#Profile after changing 'age': {'name': 'Rin', 'age': 29, 'major': 'Computer Science', 'is_enrolled': True, 'year': 'Senior'}


# 4. Removing Elements

# Using del keyword to remove an item by its key
del student_profile["is_enrolled"]
print(f"\nProfile after deleting 'is_enrolled': {student_profile}")
#Profile after deleting 'is_enrolled': {'name': 'Rin', 'age': 29, 'major': 'Computer Science', 'year': 'Senior'}

# Using .pop() method to remove an item by its key and return its value
removed_major = student_profile.pop("major")
print(f"Profile after popping 'major': {student_profile}")
print(f"Removed major was: {removed_major}")
#Profile after popping 'major': {'name': 'Rin', 'age': 29, 'year': 'Senior'}
#Removed major was: Computer Science

# Using .clear() to remove all items from the dictionary
student_profile.clear()
print(f"Profile after clearing all items: {student_profile}")#Profile after clearing all items: {}

# Let's re-create a dictionary for more examples
evening_meal = {"dumplings in chilli oil": "spicy", "icecream": "sweet", "bubble tea": "refreshing", "quantity": 3}
print(f"\nNew evening_meal dictionary: {evening_meal}")

# 5. Looping Through a Dictionary

# Loop through keys (this is the default behavior)
print("\nLooping through keys:")
for key in evening_meal:
    print(key)

# Loop through keys explicitly
print("\nLooping through keys explicitly:")
for key in evening_meal.keys():
    print(key)

# Loop through values
print("\nLooping through values:")
for value in evening_meal.values():
    print(value)

# Loop through both keys and values (items)
print("\nLooping through items (key: value):")
for key, value in evening_meal.items():
    print(f"{key}: {value}")

# 6. Checking if a Key Exists
# Use the 'in' keyword to check if a key is present in the dictionary.

print(f"\nIs 'icecream' in evening_meal dictionary? {'icecream' in evening_meal}")
print(f"Is 'french toast' not in car dictionary? {'french toast' not in evening_meal}")

# 7. Dictionary Length
# Use len() to find the number of key-value pairs.
print(f"\nNumber of items in evening_meal dictionary: {len(evening_meal)}")
```
