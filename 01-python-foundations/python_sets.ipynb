```python
print("\n--- Python Sets Example ---")

# Creating a set, for a set we use these curly brackets
my_set = {"apple", "banana", "cherry"}
print(f"Original set: {my_set}")
print(f"Type of my_set: {type(my_set)}")

# Sets do not allow duplicate values
duplicate_set = {"apple", "banana", "cherry", "apple"}
print(f"Set with duplicates (duplicates are automatically removed): {duplicate_set}")

# Adding an element to a set
my_set.add("orange")
print(f"Set after adding 'orange': {my_set}")

# Removing an element from a set
my_set.remove("banana")
print(f"Set after removing 'banana': {my_set}")

# Another way to remove, but raises an error if item is not present
# my_set.remove("grape") # This would cause an error

# Discarding an element (does not raise an error if item is not present)
my_set.discard("grape")
print(f"Set after discarding 'grape' (no change as 'grape' was not present): {my_set}")
my_set.discard("apple")
print(f"Set after discarding 'apple': {my_set}")

# Set operations
set_a = {"apple", "banana", "cherry", "date"}
set_b = {"cherry", "date", "elderberry", "fig"}

# Union (all unique elements from both sets)
union_set = set_a.union(set_b)
print(f"Union of set_a and set_b: {union_set}")

# Intersection (common elements in both sets)
intersection_set = set_a.intersection(set_b)
print(f"Intersection of set_a and set_b: {intersection_set}")

# Difference (elements in set_a but not in set_b)
difference_set = set_a.difference(set_b)
print(f"Difference (set_a - set_b): {difference_set}")

# Symmetric Difference (elements in either set, but not in both)
symmetric_difference_set = set_a.symmetric_difference(set_b)
print(f"Symmetric Difference: {symmetric_difference_set}")
```
