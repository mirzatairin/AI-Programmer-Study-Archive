```python
# A simple dictionary problem. We define a dictionary with some food items
# We find the item that costs the most.
# We add another item to the evening_meal dictionary
# We print the updated dictionary

evening_meal = {"Dumplings": 300, "Icecream": 200, "Bubbletea": 180, "Coffee": 100}

# Calculate the sum of prices
total_price = sum(evening_meal.values())

# Calculate the number of items
num_items = len(evening_meal)

# Calculate the average price
average_price = total_price / num_items

# Find the most expensive item
# The `max()` function in Python is used to find the largest item in an iterable (like a list or a dictionary).
# `key=evening_meal.get` tells `max()` to compare items based on their values, not their keys.
# `evening_meal.get` is a method that retrieves the value associated with a given key.
# So, it finds the key (food item name) whose corresponding value (price) is the largest.
most_expensive_item = max(evening_meal, key=evening_meal.get)
max_price = evening_meal[most_expensive_item]

print(f"Food Items and Prices: {evening_meal}")
print(f"Total Price: {total_price:.2f}")
print(f"Number of Items: {num_items}")
print(f"Average Price: {average_price:.2f}")
print(f"The most expensive item is '{most_expensive_item}' with a price of {max_price:.2f}")


# Part 4: Add a new model
# Add the model MobileNet with an accuracy of 0.89
new_item = 'fuchka'
new_price = 50

# Add the new model to the dictionary
evening_meal[new_item] = new_price

print(f"Updated Food Items and Prices: {evening_meal}")
```
